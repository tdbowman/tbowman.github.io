# Week 8 Walkthrough — Splitting a Program Into Modules

**Topic:** Importing modules, the standard library, writing your own

Every program so far has lived in one place. Real projects don't. We'll borrow tools from Python's standard library, then write our own module file and import it — the moment your code becomes reusable across projects instead of trapped in one notebook.

Run the cells in order. Each step adds one idea to the program, and the last
section pulls the whole thing together. Change things and re-run — that is the
whole point of a notebook.

## Step 1 — import brings in a whole module


`import math` gives you the module; you reach inside it with a dot.

```python
import math

print(math.pi)
print(math.sqrt(144))
print(math.ceil(4.2), math.floor(4.8))
```

## Step 2 — from ... import takes specific names


Shorter to type, but you lose the label that says where it came from. Use it
when the name is unambiguous.

```python
from statistics import mean, median, stdev

loans = [42, 93, 17, 61, 55, 78]

print("mean  ", round(mean(loans), 1))
print("median", median(loans))
print("stdev ", round(stdev(loans), 1))
```

## Step 3 — Aliases shorten long names


`as` renames on import. Some aliases are near-universal conventions — you'll see
`import pandas as pd` everywhere.

```python
import random as rng

rng.seed(426)                       # same seed -> same "random" numbers, every run

print(rng.randint(1, 100))
print(rng.choice(["Dune", "Beloved", "Neuromancer"]))
print(rng.sample(range(1, 50), 5))
```

## Step 4 — Three standard-library modules worth knowing now


`datetime` for dates, `random` for sampling, `statistics` for summaries. None
need installing — they ship with Python.

```python
from datetime import date, timedelta

due = date(2026, 10, 12)
today = date(2026, 10, 27)

overdue = (today - due).days
print(f"Due {due}, today {today}: {overdue} days overdue")
print("Renewal would run to", due + timedelta(days=21))
```

## Step 5 — Write your own module


A module is just a `.py` file. The `%%writefile` magic at the top of a cell
saves that cell's contents to disk instead of running it.

```python
%%writefile fines.py
"""Fine calculations for the circulation desk."""

DAILY_RATE = 0.25
MAX_FINE = 10.00


def calculate(days_overdue, daily_rate=DAILY_RATE, max_fine=MAX_FINE):
    """Return the fine owed in dollars."""
    if days_overdue <= 0:
        return 0.00
    return min(days_overdue * daily_rate, max_fine)


def describe(name, days_overdue):
    """Return a one-line notice."""
    owed = calculate(days_overdue)
    return f"{name}: on time" if owed == 0 else f"{name}: ${owed:.2f}"
```

## Step 6 — Import your own module


Same syntax as any other. Python looks in the current folder first.

```python
import fines

print(fines.DAILY_RATE)
print(fines.calculate(12))
print(fines.describe("Ana", 40))
```

## Step 7 — Editing a module you've already imported


Python caches modules. Change `fines.py` and re-importing does nothing — you
need `importlib.reload`, or just restart the kernel.

```python
import importlib
import fines

importlib.reload(fines)
print("reloaded:", fines.calculate(8))
```

## Step 8 — Third-party modules


Anything not in the standard library has to be installed first, usually with
`pip`. `pandas` is the one you'll meet in Assignment 2.

```python
try:
    import pandas as pd
    frame = pd.DataFrame({"title": ["Dune", "Beloved"], "loans": [42, 93]})
    print(frame)
except ImportError:
    print("pandas isn't installed here - run: pip install pandas")
```

## The finished program

Everything above, in one place. This is the version worth keeping.


The program is now two files: a module that knows the rules, and a script that
uses them. Change the fine policy in one place and everything follows.

```python
# Week 8 - A script that leans on modules

import random as rng
from datetime import date, timedelta
from statistics import mean

import fines

rng.seed(426)

TITLES = ["Dune", "Beloved", "Neuromancer", "Silent Spring", "Parable of the Sower"]
PATRONS = ["Ana", "Will", "Toni", "Rachel", "Octavia"]
TODAY = date(2026, 10, 27)

loans = []
for patron in PATRONS:
    title = rng.choice(TITLES)
    due = TODAY - timedelta(days=rng.randint(-5, 45))
    loans.append((patron, title, due))

print(f"CIRCULATION DESK - {TODAY}")
print("=" * 52)

owed = []
for patron, title, due in loans:
    days = (TODAY - due).days
    fine = fines.calculate(days)
    owed.append(fine)
    status = "on time" if fine == 0 else f"${fine:5.2f}"
    print(f"{patron:9} {title:22} due {due}  {status}")

print("=" * 52)
print(f"Total outstanding: ${sum(owed):.2f}")
print(f"Average fine:      ${mean(owed):.2f}")
```
