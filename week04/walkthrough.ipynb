## Step 1 — The problem: repetition


Read this and count how many times the same three lines appear.

```python
# The 'before' version - it works, but look at it
fine_ana = 12 * 0.25
if fine_ana > 10: fine_ana = 10.00
print(f"Ana owes ${fine_ana:.2f}")

fine_will = 40 * 0.25
if fine_will > 10: fine_will = 10.00
print(f"Will owes ${fine_will:.2f}")

fine_toni = 3 * 0.25
if fine_toni > 10: fine_toni = 10.00
print(f"Toni owes ${fine_toni:.2f}")
```

## Step 2 — A function names a piece of work


`def` defines it. The name should say what it *does*. Nothing runs until you
call it.

```python
def calculate_fine(days_overdue):
    return min(days_overdue * 0.25, 10.00)


print(calculate_fine(12))
print(calculate_fine(40))
print(calculate_fine(3))
```

## Step 3 — Parameters make it flexible


Hard-coded numbers inside a function are a trap. Lift them into parameters, and
give them **default values** so the common case stays short.

```python
def calculate_fine(days_overdue, daily_rate=0.25, max_fine=10.00):
    return min(days_overdue * daily_rate, max_fine)


print(calculate_fine(12))                      # uses both defaults
print(calculate_fine(12, daily_rate=0.10))     # student rate
print(calculate_fine(12, max_fine=5.00))       # lower cap
```

## Step 4 — return hands a value back; print just shows it


This trips up nearly everyone once. A function that prints gives you nothing to
work with afterwards.

```python
def shows(n):
    print(n * 0.25)          # displays, returns None

def gives(n):
    return n * 0.25          # hands the value back


a = shows(12)
b = gives(12)

print("shows returned:", a)
print("gives returned:", b)
print("and we can keep using it:", b * 2)
```

## Step 5 — Docstrings explain the contract


A string on the first line of a function is its **docstring**. `help()` reads it
back to you — and so does every editor.

```python
def calculate_fine(days_overdue, daily_rate=0.25, max_fine=10.00):
    """Return the fine owed, in dollars.

    days_overdue: whole days past the due date (0 or fewer means no fine)
    daily_rate:   charge per day
    max_fine:     ceiling, never exceeded
    """
    if days_overdue <= 0:
        return 0.00
    return min(days_overdue * daily_rate, max_fine)


help(calculate_fine)
```

## Step 6 — Scope: what a function can and can't see


Names created inside a function live only there. That is a feature — it means
you can't accidentally clobber something elsewhere.

```python
rate = 0.25          # defined out here

def show_rate():
    local_note = "only exists inside"
    print("the function can read outer names:", rate)
    print(local_note)

show_rate()

# print(local_note)   # uncomment: NameError - it's gone
```

## Step 7 — Small functions compose


Each does one thing. Together they do something useful.

```python
def calculate_fine(days, rate=0.25, cap=10.00):
    return 0.00 if days <= 0 else min(days * rate, cap)

def describe(name, days):
    fine = calculate_fine(days)
    if fine == 0:
        return f"{name}: on time"
    return f"{name}: {days} days, ${fine:.2f}"


print(describe("Ana", 12))
print(describe("Toni", 0))
```

## The finished program

Everything above, in one place. This is the version worth keeping.


Same behaviour as Step 1, a quarter of the repetition, and the fine policy now
lives in exactly one place.

```python
# Week 4 - Fines, as a small toolkit

def calculate_fine(days_overdue, daily_rate=0.25, max_fine=10.00):
    """Return the fine owed in dollars, never above max_fine."""
    if days_overdue <= 0:
        return 0.00
    return min(days_overdue * daily_rate, max_fine)


def format_notice(name, days_overdue, **rules):
    """Return a one-line notice for one patron."""
    fine = calculate_fine(days_overdue, **rules)
    if fine == 0:
        return f"{name:8} on time"
    flag = " (capped)" if fine == rules.get("max_fine", 10.00) else ""
    return f"{name:8} {days_overdue:3} days   ${fine:5.2f}{flag}"


def total_owed(records, **rules):
    """Return the sum of all fines in a list of (name, days) pairs."""
    return sum(calculate_fine(days, **rules) for _, days in records)


records = [("Ana", 12), ("Will", 40), ("Toni", 0), ("Rachel", 3)]

for name, days in records:
    print(format_notice(name, days))

print("-" * 34)
print(f"{'TOTAL':8}          ${total_owed(records):5.2f}")
```
