## Step 1 — Four types you'll use constantly


`int` whole numbers · `float` decimals · `str` text · `bool` True/False.

`type()` tells you which one you have. When something behaves strangely, this is
the first thing to check.

```python
days_overdue = 12          # int
daily_rate = 0.25          # float
item_title = "Dune"        # str
is_reference = False       # bool

for value in (days_overdue, daily_rate, item_title, is_reference):
    print(f"{str(value):10} -> {type(value).__name__}")
```

## Step 2 — Arithmetic


The usual operators, plus two worth knowing: `//` floor division (drops the
remainder) and `%` modulo (gives you *only* the remainder).

```python
days_overdue = 12
daily_rate = 0.25

print("fine       ", days_overdue * daily_rate)
print("whole weeks", days_overdue // 7)
print("leftover   ", days_overdue % 7)
```

## Step 3 — Mixing types


Python will add an int to a float happily. It will *not* add a number to a
string — and the error message is clear about why.

```python
print(3 + 0.25)             # fine: int + float

# Uncomment this line to see the error:
# print("Fine: " + 3.00)

print("Fine: " + str(3.00))  # convert first, then join
print(f"Fine: {3.00}")       # or just use an f-string
```

## Step 4 — Comparisons produce booleans


Every comparison evaluates to `True` or `False`. That's the raw material of
every decision a program makes.

```python
days_overdue = 12

print(days_overdue > 0)
print(days_overdue > 30)
print(days_overdue == 12)     # == asks a question; = makes an assignment
```

## Step 5 — Make a decision with if / elif / else


Indentation is not decoration in Python — it is how the language knows what
belongs inside the `if`.

```python
days_overdue = 12

if days_overdue == 0:
    print("Returned on time. Nothing owed.")
elif days_overdue <= 14:
    print("Overdue. A fine applies.")
else:
    print("Long overdue. Account on hold.")
```

## Step 6 — Cap the fine


Real policies have limits. `min()` gives you the smaller of two values, which is
a neat way to apply a ceiling without another `if`.

```python
days_overdue = 60
daily_rate = 0.25
max_fine = 10.00

raw_fine = days_overdue * daily_rate
fine = min(raw_fine, max_fine)

print(f"raw:    ${raw_fine:.2f}")
print(f"capped: ${fine:.2f}")
```

## The finished program

Everything above, in one place. This is the version worth keeping.


`:.2f` inside an f-string means "format as a decimal with two places" — which is
what money should always look like.

```python
# Week 2 - Late fee calculator

item_title = "Dune"
days_overdue = 23
daily_rate = 0.25
max_fine = 10.00
is_reference = False

if is_reference:
    print(f"{item_title} is reference — library use only, no loan, no fine.")
elif days_overdue <= 0:
    print(f"{item_title} was returned on time. Nothing owed.")
else:
    fine = min(days_overdue * daily_rate, max_fine)
    capped = fine == max_fine

    print(f"Item:         {item_title}")
    print(f"Days overdue: {days_overdue}")
    print(f"Fine:         ${fine:.2f}" + ("  (capped)" if capped else ""))

    if days_overdue > 30:
        print("Status:       account on hold until returned")
```
