# Week 6 Walkthrough — Ranking the Most-Borrowed Titles

**Topic:** Lists, list methods, slicing, comprehensions, tuples

Circulation data arrives as a pile of records. We'll sort it, slice out the top of the list, and build new lists from old ones — and see why some of that data should be a tuple instead.

Run the cells in order. Each step adds one idea to the program, and the last
section pulls the whole thing together. Change things and re-run — that is the

## Step 1 — A list holds an ordered collection


Square brackets. Items keep their order, and you can change them later.

```python
titles = ["Dune", "Beloved", "Neuromancer", "Silent Spring"]

print(titles)
print(titles[0])       # first
print(titles[-1])      # last
print(len(titles))
```

## Step 2 — Lists are mutable — that's the point


`append` adds to the end, `insert` puts it at a position, `remove` takes one out
by value, `pop` takes one out by position and hands it back.

```python
titles = ["Dune", "Beloved"]

titles.append("Neuromancer")
titles.insert(0, "Silent Spring")
print(titles)

dropped = titles.pop()
print("removed:", dropped)
print(titles)
```

## Step 3 — Slicing takes a section


`list[start:stop]` — same rule as strings. Leaving a side blank means "all the
way to the end".

```python
counts = [42, 17, 93, 8, 61, 25]

print(counts[:3])      # first three
print(counts[3:])      # from index 3 onward
print(counts[-2:])     # last two
print(counts[::2])     # every second item
```

## Step 4 — sorted() vs .sort()


`sorted()` returns a **new** list and leaves the original alone. `.sort()`
rearranges in place and returns `None`. Mixing them up is a classic bug.

```python
counts = [42, 17, 93, 8, 61, 25]

print("sorted() copy:", sorted(counts))
print("original:     ", counts)

counts.sort(reverse=True)
print("after .sort():", counts)
```

## Step 5 — Tuples: fixed records that shouldn't change


Round brackets. Once built, a tuple can't be modified — which makes it right for
a record whose shape is fixed, like *(title, count)*.

```python
record = ("Dune", 42)

title, count = record          # unpacking
print(title, "was borrowed", count, "times")

# record[1] = 50               # uncomment: TypeError - and that's the point
```

## Step 6 — Sort a list of tuples with a key


`key=` tells `sorted` what to compare. A tiny `lambda` is the usual way to say
"use the second item".

```python
circulation = [("Dune", 42), ("Beloved", 93), ("Neuromancer", 17), ("Silent Spring", 61)]

by_count = sorted(circulation, key=lambda pair: pair[1], reverse=True)

for title, count in by_count:
    print(f"{count:3}  {title}")
```

## Step 7 — List comprehensions build a list from a list


Read it right to left: *for each x in things, keep the ones that pass, give me
the expression.* Once it clicks it's hard to go back.

```python
circulation = [("Dune", 42), ("Beloved", 93), ("Neuromancer", 17), ("Silent Spring", 61)]

titles = [title for title, count in circulation]
popular = [title for title, count in circulation if count > 40]
shouted = [title.upper() for title, count in circulation]

print(titles)
print(popular)
print(shouted)
```

## The finished program

Everything above, in one place. This is the version worth keeping.


Tuples for the records, a list to hold them, a comprehension to filter, and a
slice to take the top three.

```python
# Week 6 - Most-borrowed report

circulation = [
    ("Dune", "Herbert", 42),
    ("Beloved", "Morrison", 93),
    ("Neuromancer", "Gibson", 17),
    ("Silent Spring", "Carson", 61),
    ("The Dispossessed", "Le Guin", 55),
    ("Parable of the Sower", "Butler", 78),
]

ranked = sorted(circulation, key=lambda record: record[2], reverse=True)
top_three = ranked[:3]
total_loans = sum(count for _, _, count in circulation)
heavy_hitters = [title for title, _, count in circulation if count >= 60]

print("TOP THREE")
print("=" * 46)
for place, (title, author, count) in enumerate(top_three, start=1):
    print(f"{place}. {title:22} {author:10} {count:3}")

print()
print(f"Titles tracked:  {len(circulation)}")
print(f"Total loans:     {total_loans}")
print(f"Average:         {total_loans / len(circulation):.1f} per title")
print(f"60 or more:      {', '.join(heavy_hitters)}")
```
