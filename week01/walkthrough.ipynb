# Week 1 Walkthrough — A Course Welcome Card

**Topic:** Variables, naming, printing, comments

We're going to build a tiny program that prints a tidy welcome card for a student. It is four lines of real Python by the end, and it introduces every habit you'll use for the rest of the semester: naming things well, commenting why, and printing something a human can read.

Run the cells in order. Each step adds one idea to the program, and the last
section pulls the whole thing together. Change things and re-run — that is the
whole point of a notebook.

## Step 1 — Print something


`print()` puts text on the screen. That's it. Text in quotes is a **string**.

```python
print("Welcome to Introduction to Programming in Python!")
```

## Step 2 — Store a value in a variable


A **variable** is a name pointing at a value. The `=` doesn't mean "equals" —
it means *"put the thing on the right into the name on the left."*

```python
student_name = "Ana"

print(student_name)
```

## Step 3 — Name things so future-you understands them


Both of these work. Only one of them still makes sense in three weeks.

Python style: lowercase words joined by underscores — `snake_case`.

```python
n = 3                       # what is n?
credit_hours = 3            # oh, that's what it is

print(n, credit_hours)
```

## Step 4 — Combine values with an f-string


Put an `f` in front of the quotes and you can drop variables straight into the
text inside `{curly braces}`.

```python
student_name = "Ana"
course_code = "INF 426"
credit_hours = 3

print(f"{student_name} is enrolled in {course_code} for {credit_hours} credits.")
```

## Step 5 — Comments explain *why*, not *what*


Anything after a `#` is ignored by Python. Good comments don't repeat the code —
they explain the reasoning a reader can't see.

```python
# Bad: says what the code already says
credit_hours = 3   # set credit hours to 3

# Good: says something the code cannot
credit_hours = 3   # standard load for graduate seminars at SOIS
```

## Step 6 — Make it look like a card


`\n` is a newline. `*` repeated builds a border. Small touches, big difference
in readability.

```python
border = "*" * 44

print(border)
print("  WELCOME")
print(border)
```

---

## The finished program

Everything above, in one place. This is the version worth keeping.


Notice how it reads top to bottom: set up the facts, then present them.

```python
# Week 1 - Course welcome card
# Prints a formatted welcome for one student.

student_name = "Ana Reyes"
course_code = "INF 426 / LIS 826"
course_title = "Introduction to Programming in Python"
credit_hours = 3
term = "Fall 2026"

border = "=" * 52

print(border)
print(f"  {course_title.upper()}")
print(f"  {course_code}  |  {term}")
print(border)
print(f"  Student:  {student_name}")
print(f"  Credits:  {credit_hours}")
print(border)
print("  Type every line of code yourself. See you Monday.")
print(border)
```
