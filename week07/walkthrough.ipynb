# Week 7 Walkthrough — A Circulation Counter

**Topic:** Dictionaries, methods, nested dictionaries, comprehensions

Lists are good at order. Dictionaries are good at *lookup* — give me the count for this title. We'll tally loans by title, then by branch, and end up with the nested structure that every API response in Block 3 will use.

Run the cells in order. Each step adds one idea to the program, and the last
section pulls the whole thing together. Change things and re-run — that is the
whole point of a notebook.

## Step 1 — Key-value pairs


Curly braces, `key: value`. Look things up by key, not by position.

```python
loans = {"Dune": 42, "Beloved": 93, "Neuromancer": 17}

print(loans["Beloved"])
print(len(loans))
print("Dune" in loans)
```

## Step 2 — Add, update, remove


Assigning to a new key creates it; assigning to an existing key replaces it.

```python
loans = {"Dune": 42}

loans["Beloved"] = 93          # new
loans["Dune"] = 45             # replaced
del loans["Beloved"]           # gone

print(loans)
```

## Step 3 — get() saves you from KeyError


`loans["Missing"]` raises. `loans.get("Missing", 0)` hands back a default. That
default is what makes counting work.

```python
loans = {"Dune": 42}

# print(loans["Missing"])          # uncomment: KeyError
print(loans.get("Missing"))        # None
print(loans.get("Missing", 0))     # 0 - much more useful
```

## Step 4 — The counting pattern


This four-line loop is one of the most useful things in Python. Learn it once,
use it forever.

```python
checkouts = ["Dune", "Beloved", "Dune", "Neuromancer", "Dune", "Beloved"]

counts = {}
for title in checkouts:
    counts[title] = counts.get(title, 0) + 1

print(counts)
```

## Step 5 — Loop over keys, values, or both


`.items()` is what you want most of the time.

```python
counts = {"Dune": 3, "Beloved": 2, "Neuromancer": 1}

for title in counts:                     # keys
    print(title, end="  ")
print()

print(sum(counts.values()))              # values

for title, n in counts.items():          # both
    print(f"{title:14} {n}")
```

## Step 6 — Sort a dictionary by its values


Dictionaries have no order you control, so sort the items into a list when you
need a ranking.

```python
counts = {"Dune": 3, "Beloved": 2, "Neuromancer": 1, "Silent Spring": 5}

ranked = sorted(counts.items(), key=lambda pair: pair[1], reverse=True)

for title, n in ranked:
    print(f"{n}  {title}")
```

## Step 7 — Nesting: a dictionary of dictionaries


This is the shape of almost every JSON response you'll meet in Week 10.

```python
branches = {
    "Main": {"Dune": 12, "Beloved": 30},
    "River Forest": {"Dune": 8, "Neuromancer": 17},
}

print(branches["Main"]["Beloved"])

for branch, holdings in branches.items():
    total = sum(holdings.values())
    print(f"{branch:14} {total:3} loans across {len(holdings)} titles")
```

## Step 8 — Dictionary comprehensions


Same idea as a list comprehension, but you produce `key: value` pairs.

```python
counts = {"Dune": 3, "Beloved": 2, "Neuromancer": 1, "Silent Spring": 5}

busy = {title: n for title, n in counts.items() if n >= 3}
doubled = {title: n * 2 for title, n in counts.items()}

print(busy)
print(doubled)
```

## The finished program

Everything above, in one place. This is the version worth keeping.


The counting pattern, a nested structure, and a sorted report — together.

```python
# Week 7 - Circulation counter

raw_checkouts = [
    ("Main", "Dune"), ("Main", "Beloved"), ("River Forest", "Dune"),
    ("Main", "Dune"), ("River Forest", "Neuromancer"), ("Main", "Beloved"),
    ("River Forest", "Dune"), ("Main", "Silent Spring"), ("Main", "Dune"),
    ("River Forest", "Beloved"), ("Main", "Neuromancer"), ("Main", "Dune"),
]

by_branch = {}
overall = {}

for branch, title in raw_checkouts:
    branch_counts = by_branch.setdefault(branch, {})
    branch_counts[title] = branch_counts.get(title, 0) + 1
    overall[title] = overall.get(title, 0) + 1

print("BY BRANCH")
print("=" * 40)
for branch, holdings in sorted(by_branch.items()):
    print(f"\n{branch}  ({sum(holdings.values())} loans)")
    for title, n in sorted(holdings.items(), key=lambda p: p[1], reverse=True):
        print(f"   {n}  {title}")

print()
print("OVERALL")
print("=" * 40)
for title, n in sorted(overall.items(), key=lambda p: p[1], reverse=True):
    share = n / len(raw_checkouts) * 100
    print(f"{n:3}  ({share:4.1f}%)  {title}")

busiest = max(overall, key=overall.get)
print(f"\nMost borrowed: {busiest}")
```
