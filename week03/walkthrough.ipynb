# Week 3 Walkthrough — Cleaning Up a Messy Name List

**Topic:** String methods, slicing, f-strings, user input

Data arrives dirty. Names come in with stray spaces, inconsistent capitals, and the surname in the wrong place. This walkthrough turns one ugly string into a clean, formatted record — the single most common chore in real data work.

Run the cells in order. Each step adds one idea to the program, and the last
section pulls the whole thing together. Change things and re-run — that is the
whole point of a notebook.

## Step 1 — The mess we're starting from


One line, badly typed by a human in a hurry.

```python
raw = "   reyes,ANA maria  "

print(repr(raw))     # repr() shows the quotes and spaces - very useful for debugging
print(len(raw))
```

## Step 2 — strip() removes the outside whitespace


Only the outside. Spaces in the middle stay put.

```python
raw = "   reyes,ANA maria  "
cleaned = raw.strip()

print(repr(cleaned))
```

## Step 3 — split() breaks a string into a list


Give it a separator and it hands back the pieces.

```python
cleaned = "reyes,ANA maria"
parts = cleaned.split(",")

print(parts)
print("surname ->", parts[0])
print("given   ->", parts[1])
```

## Step 4 — Fix the capitals


`.title()` capitalises each word. `.upper()` and `.lower()` do what they say.
None of them change the original — they hand back a *new* string.

```python
surname = "reyes"
given = "ANA maria"

print(surname.title())
print(given.title())
print(surname, "is unchanged:", surname)
```

## Step 5 — Slicing pulls out a piece by position


`text[start:stop]` — `start` is included, `stop` is not. Counting starts at 0.

```python
given = "Ana Maria"

print(given[0])        # first character
print(given[0:3])      # first three
print(given[-5:])      # last five
print(f"Initials: {given[0]}.{given.split()[1][0]}.")
```

## Step 6 — Validate before you trust


`.isdigit()`, `.isalpha()`, and the `in` keyword let you check a value before
using it. Never assume input is well-formed.

```python
def looks_like_a_name(text):
    text = text.strip()
    return len(text) > 0 and all(part.isalpha() for part in text.split())

for candidate in ("Ana Maria", "  ", "R2D2", "Reyes"):
    print(f"{candidate!r:14} -> {looks_like_a_name(candidate)}")
```

## The finished program

Everything above, in one place. This is the version worth keeping.


Each step is now one small, named function. That is a preview of Week 4.

```python
# Week 3 - Clean one patron record

def clean_name(raw):
    """Turn '  reyes,ANA maria  ' into ('Reyes', 'Ana Maria')."""
    surname, given = raw.strip().split(",")
    return surname.strip().title(), given.strip().title()


def initials(given, surname):
    first = given[0]
    last = surname[0]
    return f"{first}.{last}."


def format_record(raw, patron_id):
    surname, given = clean_name(raw)
    return (
        f"ID:       {patron_id}\n"
        f"Name:     {given} {surname}\n"
        f"Sort as:  {surname}, {given}\n"
        f"Initials: {initials(given, surname)}"
    )


messy_entries = [
    "   reyes,ANA maria  ",
    "GIBSON,william",
    " morrison ,  toni ",
]

for number, entry in enumerate(messy_entries, start=1001):
    print(format_record(entry, number))
    print("-" * 34)
```
