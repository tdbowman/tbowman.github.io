```python
# RUN THIS CELL FIRST — it defines the helper the test cells use.
# Each exercise is worth 20 points, split evenly across its checks, so getting
# four of five right earns 16 rather than 0.

def grade(title, points, checks):
    """checks: list of (label, callable_returning_bool, hint)
    hint may be a string, or a function returning a string (evaluated only
    when the check fails, so it can safely look at the student's variables)."""
    per = points / len(checks)
    earned = 0.0
    problems = []
    for label, test, hint in checks:
        try:
            ok = bool(test())
        except NameError as e:
            problems.append(f"{label}: not defined yet ({e})")
            continue
        except Exception as e:
            problems.append(f"{label}: raised {type(e).__name__}: {e}")
            continue
        if ok:
            earned += per
        else:
            try:
                message = hint() if callable(hint) else hint
            except Exception:
                message = "not correct yet"
            problems.append(f"{label}: {message}")
    earned = round(earned)
    if not problems:
        print(f"✅ {title} PASSED! {earned}/{points}")
    else:
        print(f"❌ {title}: {earned}/{points}")
        for p in problems:
            print(f"   - {p}")
    return earned

print("Grading helper loaded.")
```
