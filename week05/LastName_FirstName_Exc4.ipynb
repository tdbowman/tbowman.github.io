```python
# RUN THIS CELL FIRST — it defines the helper the test cells use.
# Each exercise gives partial credit, so a near miss is not a zero.

def grade(title, points, checks):
    """checks: list of (label, callable_returning_bool, hint)"""
    per = points / len(checks)
    earned, problems = 0.0, []
    for label, test, hint in checks:
        try:
            ok = bool(test())
        except NameError as e:
            problems.append(f"{label}: not defined yet ({e})")
            continue
        except Exception as e:
            problems.append(f"{label}: raised {type(e).__name__}: {e}")
            continue
        if ok:
            earned += per
        else:
            problems.append(f"{label}: {hint}")
    earned = round(earned)
    if not problems:
        print(f"✅ {title} PASSED! {earned}/{points}")
    else:
        print(f"❌ {title}: {earned}/{points}")
        for p in problems:
            print(f"   - {p}")
    return earned


def uses(func, *needles, forbidden=()):
    """Look at a function's own source. Returns True if every needle appears and
    no forbidden token does. If the source can't be read, returns True and says
    so — your instructor will check by eye."""
    import inspect
    try:
        src = inspect.getsource(func)
    except (OSError, TypeError):
        print("   (couldn't read the function's source — instructor will check this by eye)")
        return True
    body = "\n".join(src.split("\n")[1:])   # skip the def line
    return all(n in body for n in needles) and not any(f in body for f in forbidden)

print("Grading helper loaded.")
```
