# Week 5 Walkthrough — A Reading-Log Report

**Topic:** for loops, range, enumerate, while, break/continue, accumulators

You have a semester of reading sessions. We'll walk over them, total them up, find the best day, and stop early when a target is hit — every loop pattern you'll need, on one small dataset.

Run the cells in order. Each step adds one idea to the program, and the last
section pulls the whole thing together. Change things and re-run — that is the
whole point of a notebook.

## Step 1 — for walks over a collection


No counter to manage, no off-by-one errors. Python hands you each item in turn.

```python
pages_per_day = [12, 40, 0, 25, 60, 18, 33]

for pages in pages_per_day:
    print(pages)
```

## Step 2 — An accumulator collects a result


Start with an empty total *outside* the loop, add to it *inside*. This pattern
shows up everywhere.

```python
pages_per_day = [12, 40, 0, 25, 60, 18, 33]

total = 0
for pages in pages_per_day:
    total = total + pages          # or: total += pages

print("total pages:", total)
print("average:", total / len(pages_per_day))
```

## Step 3 — range() when you need numbers, not items


`range(7)` gives 0,1,2,3,4,5,6. `range(1, 8)` gives 1..7. The stop value is
never included.

```python
for day in range(1, 8):
    print(f"Day {day}")

print(list(range(0, 20, 5)))     # start, stop, step
```

## Step 4 — enumerate() when you need both


Reaching for `range(len(...))` to get an index is a smell. `enumerate` gives you
the position and the item together.

```python
pages_per_day = [12, 40, 0, 25, 60, 18, 33]

for day, pages in enumerate(pages_per_day, start=1):
    print(f"Day {day}: {pages:3} pages")
```

## Step 5 — Track the best as you go


Keep a running champion and replace it whenever you find better.

```python
pages_per_day = [12, 40, 0, 25, 60, 18, 33]

best_day = 0
best_pages = 0

for day, pages in enumerate(pages_per_day, start=1):
    if pages > best_pages:
        best_pages = pages
        best_day = day

print(f"Best day was day {best_day} with {best_pages} pages")
```

## Step 6 — break stops early, continue skips one


`break` leaves the loop entirely. `continue` jumps to the next item.

```python
pages_per_day = [12, 40, 0, 25, 60, 18, 33]
target = 100

running = 0
for day, pages in enumerate(pages_per_day, start=1):
    if pages == 0:
        continue                       # skip rest days, don't count them
    running += pages
    if running >= target:
        print(f"Hit {target} pages on day {day}")
        break
else:
    print(f"Never reached {target}")   # runs only if the loop was not broken
```

## Step 7 — while repeats until a condition changes


Use `while` when you don't know how many times — only when to stop. Always make
sure something inside the loop moves you toward stopping.

```python
pages_left = 320
per_session = 45
sessions = 0

while pages_left > 0:
    pages_left -= per_session
    sessions += 1

print(f"{sessions} sessions to finish the book")
```

## The finished program

Everything above, in one place. This is the version worth keeping.


One pass over the data answers four questions at once.

```python
# Week 5 - Reading log report

pages_per_day = [12, 40, 0, 25, 60, 18, 33, 0, 47, 22]
goal = 250

total = 0
active_days = 0
best_day, best_pages = 0, 0
goal_day = None

for day, pages in enumerate(pages_per_day, start=1):
    if pages == 0:
        continue

    total += pages
    active_days += 1

    if pages > best_pages:
        best_day, best_pages = day, pages

    if goal_day is None and total >= goal:
        goal_day = day

print("READING LOG")
print("=" * 34)
for day, pages in enumerate(pages_per_day, start=1):
    bar = "#" * (pages // 5)
    label = f"Day {day:2}"
    print(f"{label} {pages:3} {bar}" if pages else f"{label}   -  (rest)")

print("=" * 34)
print(f"Total pages:   {total}")
print(f"Days read:     {active_days} of {len(pages_per_day)}")
print(f"Average:       {total / active_days:.1f} per active day")
print(f"Best day:      day {best_day} ({best_pages} pages)")
print(f"Reached {goal}:  " + (f"day {goal_day}" if goal_day else "not yet"))
```
